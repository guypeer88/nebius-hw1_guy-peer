# Task 1 - Evaluation Rubric Definition

The following rubric is used to evaluate generated e-commerce product descriptions. It separates text-quality criteria from operational criteria and defines clear pass/fail and go/no-go rules.

## Evaluation Rubric

| Criterion | Good | Ok | Bad |
| - | - | - | - |
| Length | 50-90 words | 40-49 words or 91-110 words | Fewer than 40 words or more than 110 words |
| Grammar | No visible spelling, grammar, capitalization, or punctuation errors | Exactly 1 minor language or punctuation error | 2 or more language or punctuation errors, or any error that makes the text look unprofessional |
| Fluency | The description reads naturally, with clear sentence flow, no awkward word order, no unnecessary repetition, and an appropriate level of detail for the product within the target length | The description is understandable, but includes slight awkward phrasing, mild repetition, or somewhat too much or too little detail | The description feels robotic, choppy, repetitive, awkwardly ordered, or clearly over-explained or under-explained in a way that hurts readability |
| Tone | The description uses a friendly, credible, and appropriately persuasive sales voice. It is not pushy, outdated, exaggerated, or emotionally overpromising | The tone is generally suitable, but slightly bland, slightly too sales-oriented, mildly outdated, or somewhat over-polished | The tone is overly promotional, pushy, outdated, exaggerated, or includes unsupported emotional promises that reduce credibility |
| Grounding | The description stays faithful to the provided product information. Mild and reasonable marketing interpretation is allowed, as long as it does not add new factual claims, performance claims, or unsupported quality promises | The description is mostly grounded, but includes 1 mild unsupported embellishment or inference that does not materially change the meaning of the product | The description adds unsupported features, qualities, performance claims, or other information that could mislead the reader about the product |
| Latency | 0-1500 ms | 1501-3000 ms | More than 3000 ms |
| Cost | $0.0000-$0.0020 per generation | $0.0021-$0.0050 per generation | More than $0.0050 per generation |

## Pass / Fail Rules

| Rule Type | Definition |
| - | - |
| Pass | Grounding is not bad, Length is not bad, at least 3 of the 5 text-quality criteria are rated good, and no more than 1 text-quality criterion is rated bad |
| Fail | Grounding is bad, or Length is bad, or 2 or more text-quality criteria are rated bad, or fewer than 3 text-quality criteria are rated good |

## Go / No-Go Rules

| Rule | Meaning | Rationale |
| - | - | - |
| No-Go | Grounding = bad | A product description must not misrepresent the product or invent unsupported claims |
| No-Go | Length = bad | A description that is far too short or far too long does not meet the required format and hurts usability |
| No-Go | Fail according to the pass/fail threshold above | This ensures that weak descriptions are rejected even if they are factually grounded |
| Go | The description passes the pass/fail threshold above and has no no-go flags | This means the description is usable, credible, and sufficiently strong across the core quality dimensions |
