## Task 5 - LLM as a Judge

In this task, I used an LLM judge to evaluate generated product descriptions according to the rubric defined in Task 1.

The judge receives both the original product information and the generated description. This is necessary because grounding can only be evaluated by comparing the description to the source data.

The judge evaluates the following text-quality criteria:

- fluency
- grammar
- tone
- length
- grounding
- final_score

Latency and cost are not judged by the LLM, because they are measured programmatically and do not require textual evaluation.

For each criterion, the judge first provides an explanation and only then returns a verdict. This structure is intended to encourage rubric-based reasoning before assigning a label.

Run command: `uv run python -m src.judge_main`

I also ran an experiment using a larger model: `uv run python -m src.judge_main --experiment larger_model --llm-model Qwen/Qwen3-30B-A3B-Instruct-2507`,
mostly to see how the verdicts differ between the too. The small model seemed to be a little lenient around the grounding metric and it was also mistaken in evaluating the "length" score sometimes.

The larger judge model improved rubric adherence in some areas, especially length and grammar. For example, it correctly classified valid 50-90 word descriptions as GOOD and detected visible surface errors more reliably. It also caught at least one grounding problem that the smaller judge missed.

However, the larger model remained a bit lenient on grounding. It often accepted mild unsupported quality claims such as "premium", "durable", or "built to last" as fully grounded, even though these should often have reduced the grounding score to OK under the rubric. As a result, the larger judge was stronger overall, but still not fully aligned with the intended strictness of the human evaluation.
