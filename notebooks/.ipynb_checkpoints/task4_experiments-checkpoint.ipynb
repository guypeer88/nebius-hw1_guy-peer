## Task 4 - Improving the Baseline

The baseline results were already strong in grammar, tone, length, and many grounding cases.  
Because of that, I focused on small targeted experiments rather than a full redesign.

The two main opportunities for improvement were:

- mild unsupported wording in some descriptions, which affected **grounding**
- consistently poor **latency** scores

---

## Experiment 1 - Stronger grounding prompt

```bash
uv run python -m src.main --experiment grounding_v2 --take-rows 15
```

### What I changed

I rewrote the system prompt to more explicitly forbid unsupported quality claims, exaggerated wording, and emotional promises.

### Why I expected it to help

Some baseline outputs used words like "premium", "durable", or "built to last" even when those qualities were not directly supported by the product data. I expected a stricter prompt to improve grounding.

### Result

The outputs became slightly more restrained and better grounded.  
Grounding improved, while tone and fluency stayed mostly strong.

### Conclusion

This was the most useful quality-focused change.

---

## Experiment 2 - Lower temperature

### What I changed
```bash
uv run python -m src.main --experiment low_temperature --take-rows 15 --temperature 0.2   
```

I lowered the temperature to reduce randomness and make the outputs more controlled.

- Baseline - `temperature = 0.7`
- Experiment - `temperature = 0.2`

### Why I expected it to help

Lower temperature usually reduces unnecessary embellishment and makes outputs more consistent.

### Result

The outputs became more stable and slightly more grounded, but also a bit less expressive.

### Conclusion

This was a modest improvement. It helped consistency more than overall quality.

---

## Experiment 3 - Efficiency-oriented setup
```bash
uv run python -m src.main --experiment faster_model --take-rows 15 --llm-model meta-llama/Meta-Llama-3.1-8B-Instruct-fast
```

### What I changed

I tested a configuration aimed at improving latency, such as a shorter prompt and a faster model.

### Why I expected it to help

The baseline latency scores were consistently bad, so I wanted to test whether I could improve response time without hurting quality too much.

### Result

Latency improved, while text quality remained acceptable overall.  
Some outputs were slightly less polished than the baseline.

### Conclusion

This experiment improved operational performance and showed a useful quality-speed tradeoff.

---

## Overall takeaway

Since the baseline was already good, the goal of iteration was refinement, not rescue.

Overall:

- the **grounding-focused prompt** gave the best quality improvement
- the **lower-temperature setup** improved consistency
- the **efficiency-oriented setup** improved latency

These experiments showed that once the baseline is strong, the next step is focused optimization rather than major changes.
